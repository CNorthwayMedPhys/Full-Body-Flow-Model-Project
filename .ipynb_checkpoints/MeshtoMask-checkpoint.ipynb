```python
import numpy as np
import pyvista as pv
import os
import pymeshfix as mf
import PVGeo
import trimesh
import networkx as nx

###Based on https://banesullivan.com/pyvista/examples/fault_mask.html
```

```python
#Import mesh file
path = os.getcwd() + '\\VesselFiles\\Utilized Subset\\Cleaned_dias_aorta.stl'
mesh = pv.read(path)
```

```python
#Create grid onto which the binary array will be created
grid = pv.ImageData()
#Number of cells in each direction
grid.dimensions = (52,83,350)
#Bottom SW corner
grid.origin = (-11,-29,-2)
# Cell size
grid.spacing=(1,1,1)



```

```python
tmesh = trimesh.load_mesh(path)


# Fill holes in the mesh
trimesh.repair.fill_holes(tmesh)

# Repair mesh for consistency
b = trimesh.repair.broken_faces(tmesh)
trimesh.repair.stitch(tmesh, faces=b, insert_vertices= True)

mesh = pv.wrap(tmesh)
unstructured_grid = mesh.cast_to_unstructured_grid()
mask = unstructured_grid.voxelize_binary_mask()
```

```python
grid.compute_implicit_distance(mesh, inplace=True)

inner = grid.threshold(0.0, scalars='implicit_distance', invert=True)
outer = grid.threshold(0.0, scalars='implicit_distance', invert=False)

p = pv.Plotter()
p.add_mesh(mesh, color='w', label='Surface', opacity=0.25)
p.add_mesh(
    outer,
    scalars='implicit_distance',
    show_edges=True,
    opacity=0.75,
    label='Inner region',
    clim=[-1, 1],
    cmap='bwr',
)
p.add_legend()
p.show()

```

```python
import itk
MeshType = itk.Mesh[itk.F, 3]
reader = itk.MeshFileReader[MeshType].New()
meshIO = itk.STLMeshIO.New()
reader.SetMeshIO(meshIO)
reader.SetFileName(path)
reader.Update()
mesh = reader.GetOutput()
```

```python
ConverterType = itk.TriangleMeshToBinaryImageFilter[MeshType, UCType]
filter = ConverterType.New()
filter.SetInput(mesh)  # mesh is read from file
# referenceImage is read from file or constructed by setting size, origin, and spacing
filter.SetInfoImage(grid)
filter.SetInput(255)
filter.Update()
filter.GetOutput()
```

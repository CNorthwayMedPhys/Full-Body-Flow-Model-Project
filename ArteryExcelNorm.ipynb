```python
import tkinter as tk
import tkinter.filedialog as fd
import numpy as np
import easygui
import openpyxl
```

```python
#load in all excel files
tk.Tk().withdraw()
filez = fd.askopenfilenames()
```

```python
#Open one file at a time
for file in filez:
    wb = openpyxl.load_workbook(file)
    ws = wb.active
    ns = wb.copy_worksheet(ws)
    
    length_row = ws[1]
    olength=[]
    
    time_col = ws['A']
    otime = []
    
    for cell in length_row:
        olength.append(cell.value)

    for cell in time_col:
        otime.append(cell.value)
        
    #Normalize time to 0.955 
    #Determine the size of the time array and create a [0,0.955] s array of the same size
    maxtime = 0.955
    time_size = len(otime)-1
    ntime = np.append(ws['A1'].value,np.linspace(0, maxtime, num=time_size))

    #Normalize length to new length (m)
    #Ask for the new length to normalize to (cm) 
    maxlength = float(easygui.enterbox("New length for " + file+ ' in cm'))
    length_size = len(olength)-1
    nlength = np.append(ws['A1'].value,np.linspace(0, maxlength/100 , num=length_size))

    i=1
    for value in ntime:
        ws.cell(i,1).value=value
        i += 1
    i = 1
    for value in nlength:
        ws.cell(1,i).value = value
        i += 1 
        
    #Write, save, and close the excel sheet    
    wb.save(file)
    wb.close()


```

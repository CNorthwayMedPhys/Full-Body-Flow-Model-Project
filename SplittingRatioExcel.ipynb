```python
import tkinter as tk
import tkinter.filedialog as fd
import numpy as np
import easygui
import openpyxl
```

```python
#Ask for the relevant ID numbers
aID = str(easygui.enterbox('Enter vessel ID number A'))
bID = str(easygui.enterbox('Enter vessel ID number B'))

#Ask for the data
aQ = str(easygui.enterbox('Enter Q data A')) #seperated by ;
bQ = str(easygui.enterbox('Enter Q data B')) #seperated by ;

#Parse the data 
aQ = float(aQ.split(';')) #m3/s
bQ = float(aQ.split(';')) #m3/s
```

```python
#Determine the splitting ratios
aSR = aQ/(aQ+bQ)
bSR = bQ/(aQ+bQ)
name = aID+;+bID
SR = [name]

i = 0
for valueA in aSR:
    valueB = bSR[i]
    
    i += 1
    item = str(round(aSR,2)) + ; + str(round(bSR,2))
    SR.append(item)   
```

```python
#Open and write to the excel sheet
tk.Tk().withdraw()
file = fd.askopenfilename()
#Rewrite to actual file location as appropriate! 
wb = openpyxl.load_workbook(file,data_only=True)
ws = wb.worksheets[0]
col = openpyxl.utils.get_column_letter(ws.max_column)

i=1
for value in SR:
     ws.cell(i,col).value = value
    i += 1 

wb.save(file)
wb.close()
```
